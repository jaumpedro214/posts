## DuckLake First Look

💡 Run te following commands to see the ISOLATION of ACID Transactions in DuckLake in action.

```python
import duckdb
```

Create a new DuckDB client in-memmory

```python
db = duckdb.connect(':memmory:')
db.sql('''
    INSTALL ducklake;
    INSTALL postgres;
''')
```

Attach to the DuckLake Metadata store

```python
db.sql('''
    ATTACH 'ducklake:sqlite:/data/electorate/metadata.sqlite' AS electorate (DATA_PATH '/data/electorate/');
''')

db.sql('USE electorate;')
```

Select the data in `current_electorate_per_state`.

Note that the test record inserted inside the transacition is not present here.

```python
db.sql('''
    SELECT
        *
    FROM current_electorate_per_state
''')
```
